```python
import os
import json

def validate_json_schema(folder_path):
    schema = {
        "turn_number": int,
        "speaker": str,
        "role": str,
        "content": str
    }
    
    allowed_roles = [
        "Democratic Presidential Candidate", 
        "Republican Presidential Candidate", 
        "Independent Presidential Candidate", 
        "Moderator"
    ]

    def validate_record(record):
        errors = []
        
        # Check if all required keys are present
        for key, value_type in schema.items():
            if key not in record:
                errors.append(f"Missing key '{key}'")
            elif not isinstance(record[key], value_type):
                errors.append(f"Incorrect type for key '{key}': expected {value_type.__name__}, got {type(record[key]).__name__}")
        
        # Validate 'role' value
        if record.get("role") not in allowed_roles:
            errors.append(f"Invalid role '{record.get('role')}'")
        
        return errors

    def validate_file(file_path):
        with open(file_path, 'r', encoding='utf-8') as f:
            try:
                data = json.load(f)
            except json.JSONDecodeError as e:
                print(f"File '{file_path}' is not valid JSON. Error: {e}")
                return
            except UnicodeDecodeError as e:
                print(f"File '{file_path}' has encoding issues. Error: {e}")
                return
            
            if not isinstance(data, list):
                print(f"File '{file_path}' does not contain a list of records.")
                return

            print(file_path)
            
            candidates = {
                "Democratic Presidential Candidate": set(),
                "Republican Presidential Candidate": set(),
                "Independent Presidential Candidate": set()
            }
            
            for i, record in enumerate(data):
                errors = validate_record(record)
                if errors:
                    print(f"Errors in file '{file_path}', record {i}:")
                    for error in errors:
                        print(f"  - {error}")
                
                role = record.get("role")
                speaker = record.get("speaker")
                
                if role in candidates:
                    candidates[role].add(speaker)
            
            for role, speakers in candidates.items():
                if len(speakers) > 1:
                    print(f"File '{file_path}' has multiple speakers for the role '{role}': {', '.join(speakers)}. Expected exactly 1.")
                elif len(speakers) == 0 and role != "Independent Presidential Candidate":
                    print(f"File '{file_path}' has no speaker for the role '{role}'. Expected exactly 1.")
    
    for root, _, files in os.walk(folder_path):
        for file in files:
            if file.endswith(".json"):
                validate_file(os.path.join(root, file))
```

```python
validate_json_schema('transcripts_json')
```
