Open transcript and add turn number

```python
import json

def read_json_file(file_path):
    encodings = ['utf-8', 'utf-8-sig', 'latin-1']
    for enc in encodings:
        try:
            with open(file_path, 'r', encoding=enc) as file:
                return json.load(file)
        except UnicodeDecodeError:
            continue
    raise ValueError(f"Unable to decode file with any of the tried encodings: {encodings}")

def write_json_file(file_path, data):
    with open(file_path, 'w', encoding='utf-8') as file:
        json.dump(data, file, indent=2, ensure_ascii=False)

def add_turns(data):
    for i, entry in enumerate(data, start=1):
        entry["turn_number"] = i
    return data

# File path for the JSON file
file_path = 'transcript.json'

# Read the JSON data from the file
data = read_json_file(file_path)

# Add turns to data
data_with_turns = add_turns(data)

# Save the modified data back to the same JSON file
write_json_file(file_path, data_with_turns)
```
